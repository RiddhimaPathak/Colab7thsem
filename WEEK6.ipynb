```python
from sklearn.datasets import fetch_california_housing

# Load dataset
housing = fetch_california_housing(as_frame=True)
housing_df = housing.frame

# Print original data
print("California Housing Data (first 5 rows):\n")
print(housing_df.head())

print("\nOriginal columns:\n")
print(housing_df.columns)

# Select required features
selected_columns = [
    "MedInc",
    "HouseAge",
    "AveRooms",
    "AveBedrms",
    "Population"
]

# Create new DataFrame
selected_features_df = housing_df[selected_columns]

# Print selected data
print("\nSelected Features Data (first 5 rows):\n")
print(selected_features_df.head())

# Print selected columns
print("\nSelected columns:\n")
print(selected_features_df.columns)
```

DISCUSSION:
Reducing the number of features makes a machine learning model simpler, faster, and easier to interpret. It can also reduce overfitting and noise. However, removing too many features may cause the model to lose useful information and reduce its prediction accuracy.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Load the Iris dataset
iris = load_iris(as_frame=True)

# Store the data in a DataFrame
iris_df = iris.frame

# Select the four numerical features
features = [
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)",
    "petal width (cm)"
]

# -------------------------------
# 1. Min-Max Normalization -> It transforms every feature so that its values are
# generally between 0 and 1.
# -------------------------------

minmax_scaler = MinMaxScaler()

normalized_data = minmax_scaler.fit_transform(iris_df[features])

# Create a DataFrame with normalized values
normalized_df = iris_df[features].copy()
normalized_df[features] = normalized_data

print("Normalized Data (first 5 rows):")
print(normalized_df.head())


# -------------------------------
# 2. Standardization
# -------------------------------

standard_scaler = StandardScaler()

standardized_data = standard_scaler.fit_transform(iris_df[features])

# Create a DataFrame with standardized values
standardized_df = iris_df[features].copy()
standardized_df[features] = standardized_data

print("\nStandardized Data (first 5 rows):")
print(standardized_df.head())

```

DISCUSSION
Normalization scales data to a fixed range, usually 0 to 1, while standardization transforms data so that it has a mean of approximately 0 and a standard deviation of 1. Normalization is useful when a fixed range is required, whereas standardization is useful when features have different scales and we want them centered around zero. Both techniques help machine learning algorithms work better when features have very different numerical scales.

```python
import pandas as pd

# Dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"

# Column names
column_names = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race',
    'sex', 'capital-gain', 'capital-loss', 'hours-per-week',
    'native-country', 'income'
]

# Load dataset directly from the internet
adult_df = pd.read_csv(
    url,
    header=None,
    names=column_names,
    skipinitialspace=True
)

# Display first 5 rows
print("Adult Income Data (first 5 rows):")
print(adult_df.head())


# Discretization
bins = [17, 30, 45, 60, adult_df['age'].max()]

labels = [
    'Young Adult',
    'Middle-Aged',
    'Senior Adult',
    'Elderly'
]

adult_df['age_group'] = pd.cut(
    adult_df['age'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Display age group counts
print("\nAge Group Counts:")
print(adult_df['age_group'].value_counts())
```

```python
import pandas as pd
import zipfile
import urllib.request

# Download MovieLens 100K dataset
url = "https://files.grouplens.org/datasets/movielens/ml-100k.zip"
urllib.request.urlretrieve(url, "ml-100k.zip")

# Extract the ZIP file
with zipfile.ZipFile("ml-100k.zip", "r") as zip_ref:
    zip_ref.extractall(".")

# -------------------------------
# Load ratings data (u.data)
# -------------------------------

ratings_df = pd.read_csv(
    "ml-100k/u.data",
    sep="\t",
    names=["user_id", "item_id", "rating", "timestamp"]
)

# -------------------------------
# Load movie data (u.item)
# -------------------------------

movies_df = pd.read_csv(
    "ml-100k/u.item",
    sep="|",
    encoding="latin-1",
    header=None,
    usecols=[0, 1],
    names=["item_id", "movie_title"]
)

# -------------------------------
# Calculate movie statistics
# -------------------------------

movie_stats = ratings_df.groupby("item_id").agg(
    avg_rating=("rating", "mean"),
    num_ratings=("rating", "count")
).reset_index()

# -------------------------------
# Merge ratings statistics
# with movie titles
# -------------------------------

movie_stats_df = pd.merge(
    movies_df,
    movie_stats,
    on="item_id"
)

# Keep only required columns
movie_stats_df = movie_stats_df[
    ["movie_title", "avg_rating", "num_ratings"]
]

# Sort by number of ratings
movie_stats_df = movie_stats_df.sort_values(
    "num_ratings",
    ascending=False
)

# Display top 10 movies
print("Top 10 Movies by Number of Ratings:")
print(movie_stats_df.head(10))
```
