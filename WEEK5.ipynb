```python
import pandas as pd
import numpy as np


np.random.seed(42)

performance_scores = np.random.normal(
    loc=75,
    scale=10,
    size=100
)

# Add outliers
performance_scores[10] = 150   # Extreme high outlier
performance_scores[25] = -20   # Extreme low outlier
performance_scores[50] = 5     # Less extreme low outlier

# Create DataFrame
employee_df = pd.DataFrame({
    'EmployeeID': range(1, 101),
    'PerformanceScore': performance_scores.round()
})

# Display original data
print("Original Employee Performance Data (first 10 rows):\n")
print(employee_df.head(10))

print("\nOriginal Employee Performance Data (last 10 rows):\n")
print(employee_df.tail(10))


Q1 = employee_df['PerformanceScore'].quantile(0.25)
Q3 = employee_df['PerformanceScore'].quantile(0.75)

IQR = Q3 - Q1

print("\n----------------------------------------")
print("OUTLIER DETECTION USING IQR METHOD")
print("----------------------------------------")

print("Q1 (25th percentile):", Q1)
print("Q3 (75th percentile):", Q3)
print("IQR:", IQR)



lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)


outlier_condition = (
    (employee_df['PerformanceScore'] < lower_bound) |
    (employee_df['PerformanceScore'] > upper_bound)
)

outliers = employee_df[outlier_condition]

print("\n----------------------------------------")
print("OUTLIERS")
print("----------------------------------------")

print(outliers[['EmployeeID', 'PerformanceScore']])


# Replace values above upper bound with upper bound
employee_df.loc[
    employee_df['PerformanceScore'] > upper_bound,
    'PerformanceScore'
] = upper_bound

# Replace values below lower bound with lower bound
employee_df.loc[
    employee_df['PerformanceScore'] < lower_bound,
    'PerformanceScore'
] = lower_bound


print("\n----------------------------------------")
print("PERFORMANCE SCORES AFTER CAPPING")
print("----------------------------------------")

print(employee_df[['EmployeeID', 'PerformanceScore']].to_string(index=False))



remaining_outliers = employee_df[
    (employee_df['PerformanceScore'] < lower_bound) |
    (employee_df['PerformanceScore'] > upper_bound)
]

print("\n----------------------------------------")
print("REMAINING OUTLIERS AFTER CAPPING")
print("----------------------------------------")

print(remaining_outliers[['EmployeeID', 'PerformanceScore']])
```

```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# Activity 3: Integrating Global Temperature and City Data
# ---------------------------------------------------------

np.random.seed(43)



temp_data = pd.DataFrame({
    'CityName': [
        'London', 'Paris', 'New York', 'Tokyo',
        'Berlin', 'Rome', 'Sydney', 'Cairo'
    ],
    'AvgTempC': np.random.uniform(10, 30, 8).round(1),
    'Humidity': np.random.uniform(50, 90, 8).round(1)
})

print("\nTemperature Data:\n")
print(temp_data)




city_pop_data = pd.DataFrame({
    'City': [
        'London', 'Paris', 'New York', 'Tokyo',
        'Sydney', 'Dubai', 'Beijing'
    ],
    'Country': [
        'UK', 'France', 'USA', 'Japan',
        'Australia', 'UAE', 'China'
    ],
    'PopulationMillions': [
        9, 2.1, 8.4, 14, 5.3, 3.1, 21.5
    ]
})

print("\nCity Population Data:\n")
print(city_pop_data)



integrated_data = pd.merge(
    temp_data,
    city_pop_data,
    left_on='CityName',
    right_on='City',
    how='inner'
)


print("\nIntegrated DataFrame:\n")
print(integrated_data)
```

```python
df=pd.read_csv("house_prices_train.csv")
print("Original Dataset Shape:", df.shape)
print(df.head())
```

```python
# Step 3: Identify missing values
print("\nMissing Values in Each Column:")
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))

# Step 4: Drop columns having more than 40% missing values
threshold = len(df) * 0.40
df = df.dropna(axis=1, thresh=threshold)

print("\nDataset Shape After Dropping Columns:")
print(df.shape)

# Step 5: Identify numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

# Step 6: Fill missing numerical values with median
for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())

# Step 7: Identify categorical columns
categorical_columns = df.select_dtypes(include="object").columns

# Step 8: Fill missing categorical values with mode
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Step 9: Verify remaining missing values
print("\nTotal Remaining Missing Values:")
print(df.isnull().sum().sum())

# Step 10: Remove duplicate rows
df = df.drop_duplicates()

print("\nDataset Shape After Removing Duplicates:")
print(df.shape)

# Step 11: Detect and cap outliers using IQR
for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap values above upper bound
    df.loc[df[col] > upper_bound, col] = upper_bound

    # Cap values below lower bound
    df.loc[df[col] < lower_bound, col] = lower_bound

print("\nOutliers Capped Successfully!")

# Step 12: Convert categorical columns into numerical columns
df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True
)

print("\nDataset Shape After One-Hot Encoding:")
print(df.shape)

# Step 13: Standardize numerical columns
scaler = StandardScaler()

df[numerical_columns] = scaler.fit_transform(
    df[numerical_columns]
)

print("\nNumerical Features Standardized Successfully!")

# Step 14: Display final cleaned dataset
print("\nFinal Cleaned Dataset Shape:")
print(df.shape)

print("\nFirst 5 Rows of Final Dataset:")
print(df.head())

# Step 15: Save the cleaned dataset
df.to_csv("HousePrices_Cleaned.csv", index=False)

print("\nCleaned dataset saved as:")
print("HousePrices_Cleaned.csv")
```
